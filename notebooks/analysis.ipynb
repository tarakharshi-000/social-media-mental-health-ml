# Social Media Mental Health Impact Analysis

This notebook contains the complete analysis for predicting mental health impact from social media usage patterns.

## Research Question
Can we predict whether users perceive social media as affecting their mental health based on their usage patterns and behavior?

## Paper Reference
Laila, A. B. Z. (2026). Detecting the Impact of Social Media on Users' Mental Health Using Machine Learning and XAI. *Jurnal Buana Informatika, 17*(1).

```python
# Import necessary libraries
import sys
sys.path.append('../src')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import our modules
from data_preprocessing import full_preprocessing_pipeline
from train import train_all_models, evaluate_all_models, perform_cv_all_models
from evaluate import save_all_results
from additional_analysis import perform_additional_analysis

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

print("Libraries imported successfully!")
```

## 1. Data Loading and Preprocessing

```python
# Load and preprocess data
file_path = r'../data/Fai Datasets Rishu.xlsx'
processed_data = full_preprocessing_pipeline(file_path)

# Extract processed data
X_train = processed_data['X_train']
X_test = processed_data['X_test']
y_train = processed_data['y_train']
y_test = processed_data['y_test']
feature_names = processed_data['feature_names']

print(f"Training set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
print(f"Number of features: {len(feature_names)}")
```

## 2. Model Training

```python
# Train all models
models = train_all_models(X_train, y_train)

print("\nModels trained:")
for key in models.keys():
    print(f"  - {key}")
```

## 3. Model Evaluation

```python
# Evaluate all models
results = evaluate_all_models(models, X_test, y_test)

# Perform cross-validation
cv_results = perform_cv_all_models(models, X_train, y_train)
```

## 4. Results Summary

```python
# Create results summary
results_summary = []

model_names = {
    'xgboost': 'XGBoost',
    'random_forest': 'Random Forest',
    'naive_bayes': 'Naïve Bayes',
    'baseline': 'Baseline'
}

for key, name in model_names.items():
    if key in results:
        summary = {
            'Model': name,
            'Test Accuracy': f"{results[key]['accuracy']:.4f}",
            'Test Precision': f"{results[key]['precision']:.4f}",
            'Test Recall': f"{results[key]['recall']:.4f}",
            'Test F1-Score': f"{results[key]['f1_score']:.4f}",
            'CV Mean Accuracy': f"{cv_results[key]['cv_mean']:.4f}",
            'CV Std Accuracy': f"{cv_results[key]['cv_std']:.4f}"
        }
        results_summary.append(summary)

results_df = pd.DataFrame(results_summary)
results_df
```

## 5. Save Results

```python
# Save all results
results_df, paper_comparison_df = save_all_results(results, cv_results, save_dir='../results')

print("\nResults saved successfully!")
```

## 6. Additional Analysis

```python
# Perform additional analysis
perform_additional_analysis(models, X_train, X_test, y_train, y_test, feature_names, save_dir='../results')

print("\nAdditional analysis completed!")
```
